```python
import minimalmodbus

# 1. Настройка порта и адреса устройства (slaveaddress=18)
# Замените 'COM5' на ваш реальный номер порта из Modbus Poll
PORT_NAME = r'\\.\COM3'  # Префикс \\.\ решает проблемы с портами выше COM9

try:
    # Инициализируем устройство
    instrument = minimalmodbus.Instrument(PORT_NAME, slaveaddress=18)
    
    # 2. Настройка параметров связи (выставите точно как в Modbus Poll)
    instrument.serial.baudrate = 9600          # Скорость
    instrument.serial.bytesize = 8             # Биты данных
    instrument.serial.parity   = minimalmodbus.serial.PARITY_NONE  # Без четности ('E' - Even, 'O' - Odd)
    instrument.serial.stopbits = 1             # Стоповые биты
    instrument.serial.timeout  = 1.0           # Тайм-аут ожидания ответа (в секундах)
    
    # Режим работы: RTU (не ASCII)
    instrument.mode = minimalmodbus.MODE_RTU   
    
    # Очищать буфер перед каждым запросом (помогает от зависаний)
    instrument.clear_buffers_before_each_transaction = True

    print(f"Попытка опроса устройства 18 через порт {PORT_NAME}...")

    # 3. Чтение регистра 201
    # functioncode=3 означает Read Holding Registers
    # numberOfDecimals=0 означает, что читаем целое число (без деления на 10 или 100)
#register_value = instrument.read_register(201, number_of_decimals=0, functioncode=3)
    # Читаем 32-битное число, начиная с регистра 201
    # byteorder=2 включает режим Little Endian (младшее слово/байты вперед)
    # functioncode=3 означает чтение Holding Registers
    register_value = instrument.read_long(
    registeraddress=201, 
    byteorder=1, 
    signed=False, 
    functioncode=3
)
    print("--- Успешное подключение и чтение! ---")
    print(f"Значение в регистре 201: {register_value}")

except minimalmodbus.NoResponseError:
    print("Ошибка: Устройство №18 не ответило. Проверьте провода A/B, скорость и адрес.")
except minimalmodbus.InvalidResponseError:
    print("Ошибка: Получен искаженный ответ. Возможно, помехи в линии или неверная четность (parity).")
except Exception as e:
    print(f"Системная ошибка (возможно, порт всё ещё занят): {e}")
```
