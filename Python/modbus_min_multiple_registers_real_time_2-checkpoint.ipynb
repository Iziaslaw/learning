```python
import tkinter as tk
from tkinter import ttk
import threading
import time
import minimalmodbus

# --- НАСТРОЙКИ ПОДКЛЮЧЕНИЯ ---
PORT_NAME = r'\\.\COM3'
SLAVE_ADDRESS = 18
BAUDRATE = 9600
STOPBITS = 1  # Измените на 1, если потребуется

# --- КАРТА РЕГИСТРОВ (25 штук) ---
# Название: (Физический адрес, тип данных)
# Вы можете изменить названия и типы под ваши реальные датчики
REGISTERS_MAP = {
    #f"Параметр {i}": (200 + i, "uint16") for i in range(1, 26)
    'work_mode':         (1, 'uint16'),
    'Key_status_mask':         (2, 'uint16'),
    'Air_flow_counter':         (3, 'uint16'),
    'Work_Time': (201, 'uint32_le'),     # 32-bit Unsigned Little Endian
    'Time_before_next_phase': (203, 'uint32_le'),
    'Time_before_air': (205, 'uint32_le'),
    'Time_before_probe': (207, 'uint32_le'),
    'CO, ppm':      (301, 'float_le'),         # Добавьте свои регистры сюда
    'NO, ppm':      (303, 'float_le'),         # Добавьте свои регистры сюда
    'O2, %':      (305, 'float_le'),
    'tok_in0':         (307, 'float_le'),
    'tok_in1':         (309, 'float_le'),
    'calc_conc_NOx':         (311, 'float_le'),
    'calc_conc_CO2':         (313, 'float_le'),
    'calc_val_Alfa':         (315, 'float_le'),
    'calc_val_KPD_max':         (317, 'float_le'),
    'calc_val_KPD_min':         (319, 'float_le'),
    'tok_out0':      (321, 'float_le'),
    'tok_out1':      (323, 'float_le'),# Добавьте свои регистры сюда
    'tok_out2':         (325, 'float_le'),
    'Type_Protocol':         (401, 'uint16'),
    'Address':         (402, 'uint16'),
    'ID':         (501, 'uint16'),
    'Total_Work_Time': (601, 'uint32_le'),     # 32-bit Unsigned Little Endian
    'Koeff_NOx': (701, 'float_le'),
    "Авария: Перегрев":            (2, ("bit", 0)),
    "Авария: Короткое замыкание":   (2, ("bit", 1)),
    "Статус: Двигатель запущен":   (2, ("bit", 2)),
    "Статус: Дистанц. управление": (2, ("bit", 3)),
}
# Пример ручной замены конкретного регистра на ваш рабочий 32-битный:
#REGISTERS_MAP["Счетчик импульсов (32-bit)"] = (201, "uint32_le")
STATUS_BITS_MAP = {
    0: "Авария: Перегрев",
    1: "Авария: Короткое замыкание",
    2: "Статус: Двигатель запущен",
    3: "Статус: Дистанционное управление",
    # и так далее до 15 бита...
}
    
class ModbusApp(tk.Tk):
    def __init__(self):
        super().__init__()
        
        self.title(f"Мониторинг RS-485 (Адрес {SLAVE_ADDRESS})")
        self.geometry("600x650")
        self.configure(bg="#f0f0f0")
        
        self.is_running = False
        self.instrument = None
        
        self.create_widgets()
        self.protocol("WM_DELETE_WINDOW", self.on_closing)

    def create_widgets(self):
        # Верхняя панель управления
        control_frame = tk.Frame(self, bg="#e0e0e0", bd=1, relief=tk.RAISED)
        control_frame.pack(fill=tk.X, padx=10, pady=5)
        
        self.btn_toggle = tk.Button(control_frame, text="СТАРТ ОПРОСА", font=("Arial", 11, "bold"), 
                                    bg="#4CAF50", fg="white", width=15, command=self.toggle_polling)
        self.btn_toggle.pack(side=tk.LEFT, padx=10, pady=5)
        
        self.lbl_status = tk.Label(control_frame, text="Статус: Остановлен", font=("Arial", 10), bg="#e0e0e0")
        self.lbl_status.pack(side=tk.RIGHT, padx=10, pady=5)
        # --- БЛОК ДЛЯ ТЕСТИРОВАНИЯ ЗАПИСИ ---
        # Поле ввода значения
        self.entry_test_val = tk.Entry(control_frame, width=5, font=("Arial", 11))
        self.entry_test_val.insert(0, "1") # Значение по умолчанию
        self.entry_test_val.pack(side=tk.LEFT, padx=5, pady=5)
        
        # Кнопка записи
        btn_write_test = tk.Label(control_frame, text="Запись в маск.сос.ключей", font=("Arial", 9),
                                   bg="#2196F3")
        btn_write_test.pack(side=tk.LEFT, padx=5, pady=5)
        # ------------------------------------
        # Переменная для хранения состояния галочки (0 или 1)
        self.emulate_var = tk.IntVar(value=0) 

        chk_emulate = tk.Checkbutton(control_frame, text="Эмул", 
                                     variable=self.emulate_var, font=("Arial", 9), bg="#e0e0e0")
        chk_emulate.pack(side=tk.LEFT, padx=5, pady=5)
        # Таблица данных
        table_frame = tk.Frame(self)
        table_frame.pack(fill=tk.BOTH, expand=True, padx=10, pady=5)        
        
        style = ttk.Style()
        style.theme_use('clam')
        style.configure("Treeview", background="#ffffff", fieldbackground="#ffffff", rowheight=25)
        
        # Переопределяем цвета выделения (выделенная строка не будет синеть)
        style.map("Treeview", 
            background=[("selected", "invalid")], # Сбрасываем дефолтное системное выделение фона
            foreground=[("selected", "invalid")]  # Сбрасываем дефолтное системное выделение текста
        )

        columns = ("name", "address", "type", "value")
        self.tree = ttk.Treeview(table_frame, columns=columns, show="headings", selectmode="browse")

        
        self.tree.heading("name", text="Название параметра")
        self.tree.heading("address", text="Регистр")
        self.tree.heading("type", text="Тип данных")
        self.tree.heading("value", text="Значение")
        
        self.tree.column("name", width=220, anchor=tk.W)
        self.tree.column("address", width=70, anchor=tk.CENTER)
        self.tree.column("type", width=90, anchor=tk.CENTER)
        self.tree.column("value", width=180, anchor=tk.CENTER)
        
        # Стили для подсветки ошибок/успеха
        self.tree.tag_configure("success", foreground="green", background="#ffffff")
        self.tree.tag_configure("error", foreground="gray", background="#f0f0f0")
        
        # Для аварии жестко прописываем поведение: и в обычном, и в выделенном состоянии она КРАСНАЯ
        self.tree.tag_configure("alarm", foreground="white", background="#b30000")
        
        # Заполняем таблицу начальными данными
        for name, (addr, dtype) in REGISTERS_MAP.items():
            self.tree.insert("", tk.END, iid=name, values=(name, addr, dtype, "---"))
            
        scrollbar = ttk.Scrollbar(table_frame, orient=tk.VERTICAL, command=self.tree.yview)
        self.tree.configure(yscrollcommand=scrollbar.set)
        
        self.tree.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)
        scrollbar.pack(side=tk.RIGHT, fill=tk.Y)

    def init_modbus(self):
        """Безопасная инициализация порта"""
        try:
            self.instrument = minimalmodbus.Instrument(PORT_NAME, slaveaddress=SLAVE_ADDRESS)
            self.instrument.serial.baudrate = BAUDRATE
            self.instrument.serial.bytesize = 8
            self.instrument.serial.parity = minimalmodbus.serial.PARITY_NONE
            self.instrument.serial.stopbits = STOPBITS
            self.instrument.serial.timeout = 0.5  # Уменьшили тайм-аут для отзывчивости интерфейса
            self.instrument.mode = minimalmodbus.MODE_RTU
            self.instrument.clear_buffers_before_each_transaction = True
            return True
        except Exception as e:
            self.lbl_status.config(text=f"Ошибка порта: {type(e).__name__}", fg="red")
            return False

    def toggle_polling(self):
        if not self.is_running:
            # Запуск опроса
            if self.init_modbus():
                self.is_running = True
                self.btn_toggle.config(text="СТОП ОПРОСА", bg="#f44336")
                self.lbl_status.config(text="Статус: Опрос идет...", fg="green")
                
                # Запускаем поток опроса, чтобы интерфейс не зависал
                self.thread = threading.Thread(target=self.polling_loop, daemon=True)
                self.thread.start()
        else:
            # Останавливаем опрос
            self.stop_polling_process()
    def write_test_value(self):
        """Метод считывает число из текстового поля и отправляет его в регистр 2"""
        if not self.instrument:
            self.lbl_status.config(text="Ошибка: Сначала запустите опрос!", fg="red")
            return
            
        try:
            # Получаем значение из поля ввода и переводим в int
            val_to_write = int(self.entry_test_val.get())
            
            # Функция 06 Modbus (Запись одного Holding Register)
            # Отправляем значение в регистр 205
            self.instrument.write_register(205, value=val_to_write, functioncode=6)
            
            self.lbl_status.config(text=f"В рег. 205 успешно записано: {val_to_write}", fg="blue")
        except ValueError:
            self.lbl_status.config(text="Ошибка: Введите целое число!", fg="red")
        except Exception as e:
            self.lbl_status.config(text=f"Ошибка записи: {type(e).__name__}", fg="red")

    def stop_polling_process(self):
        self.is_running = False
        if self.instrument and self.instrument.serial.is_open:
            self.instrument.serial.close() # Явно закрываем порт для ОС
        self.btn_toggle.config(text="СТАРТ ОПРОСА", bg="#4CAF50")
        self.lbl_status.config(text="Статус: Остановлен", fg="black")

    def polling_loop(self):
        """Функция работает в отдельном фоновом потоке"""
        while self.is_running:
            cached_registers = {}
            for name, (address, data_type) in REGISTERS_MAP.items():
                if not self.is_running: 
                    break
                try:    
                    tag = "success"
                    # НАДЕЖНАЯ ПРОВЕРКА: Если тип данных — это кортеж бита
                    if isinstance(data_type, tuple) and data_type[0] == "bit":
                        bit_index = data_type[1] # Напрямую забираем цифру бита
                        
                        if address not in cached_registers:
                            if self.emulate_var.get() == 1 and address == 2:
                                try:
                                     cached_registers[address] = int(self.entry_test_val.get())
                                except ValueError:
                                     cached_registers[address] = 0
                            else:
                             # В обычном режиме читаем физическое устройство из сети
                             cached_registers[address] = self.instrument.read_register(address, signed=False)
                        
                        raw_reg = cached_registers[address]
                        
                        # Выделяем строго один бит математически
                        bit_value = (raw_reg >> bit_index) & 1
                        
                        if bit_value == 1:
                            val = "АКТИВНА (1)"
                            tag = "alarm" if "Авария" in name else "success"
                        else:
                            val = "Норма (0)" if "Авария" in name else "ОТКЛ (0)"
                            tag = "success"
                    # Чтение в зависимости от типа
                    elif data_type == 'uint16':
                        val = self.instrument.read_register(address, number_of_decimals=0, signed=False)
                    elif data_type == 'uint32_le':
                        val = self.instrument.read_long(address, byteorder=3, signed=False)
                    elif data_type == 'float_le':
                        val = self.instrument.read_float(address, functioncode=3, byteorder=3)
                        val = f"{val:.2f}"
                    elif data_type == 'mine':
                        val = 5 #self.instrument.read_register(address, number_of_decimals=0, signed=False)    
                    elif data_type == 'uchar_low':
                    # Читаем регистр целиком и забираем только младшие 8 бит
                        raw_reg = self.instrument.read_register(address, signed=False)
                        val = raw_reg & 0xFF  # Битовая маска                        
                    elif data_type == 'uchar_high':
                        # Читаем регистр целиком, сдвигаем на 8 бит вправо и забираем остаток
                        raw_reg = self.instrument.read_register(address, signed=False)
                        val = (raw_reg >> 8) & 0xFF  # Сдвиг и маска
                        
                    else:
                        val = self.instrument.read_register(address, signed=True)
            # Если значение является дробным числом (float), округляем его до .2f
                    if isinstance(val, float):
                        val = f"{val:.2f}"
                    # Обновляем строку в графическом интерфейсе (безопасно через root)
                    self.after(0, self.update_tree_row, name, val, tag)
                    
                except Exception as e:
                    err_msg = f"Ошибка ({type(e).__name__})"
                    self.after(0, self.update_tree_row, name, err_msg, "error")
                
                time.sleep(0.02) # Пауза между регистрами
                
            time.sleep(1.0) # Пауза между полными циклами опроса таблицы

    def update_tree_row(self, row_id, value, tag):
        """Обновление данных в таблице Tkinter"""
        if self.tree.exists(row_id):
            current_vals = self.tree.item(row_id, "values")
            self.tree.item(row_id, values=(current_vals[0], current_vals[1], current_vals[2], value), tags=(tag,))

    def on_closing(self):
        """Действие при закрытии окна крестиком"""
        self.stop_polling_process()
        self.destroy()


# Запуск приложения
if __name__ == "__main__":
    app = ModbusApp()
    app.mainloop()
```
